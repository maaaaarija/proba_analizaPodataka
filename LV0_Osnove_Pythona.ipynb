## 1. Osnovna sintaksa i tipovi podataka

- `print()` ispisuje na ekran  
- Tipovi podataka: `int`, `float`, `str`, `bool`  
- Pretvorbe tipova: `int()`, `float()`, `str()`  
- Aritmetičke i logičke operacije


````python

# Ispis osnovne poruke
print("Pozdrav, Python!")

# Primjeri tipova podataka
a = 5              # int
b = 3.14           # float
ime = "Ana"        # str
aktivan = True     # bool

# Pretvorbe tipova
a_kao_float = float(a)
b_kao_str = str(b)

print("a =", a, type(a))
print("b =", b, type(b))
print("ime =", ime, type(ime))
print("aktivan =", aktivan, type(aktivan))
print("a_kao_float =", a_kao_float, type(a_kao_float))
print("b_kao_str =", b_kao_str, type(b_kao_str))

# Aritmetika
zbroj = a + 2
kvadrat = a ** 2
cjelobrojno = 7 // 3
ostatak = 7 % 3

print("zbroj:", zbroj, "| kvadrat:", kvadrat, "| cjelobrojno:", cjelobrojno, "| ostatak:", ostatak)

# Logika
print("Primjer logike:", (a > 3) and (b < 4))
````


## 2. Strukture podataka

- **Lista** (promjenjiva, uređena)  
- **n-torka** (nepromjenjiva, uređena)  
- **Skup** (jedinstveni elementi, neuređeno)  
- **Rječnik** (ključ–vrijednost)  
- Petlja `for` i *list comprehension*


````python

# Lista
brojevi = [1, 2, 3, 4]
brojevi.append(5)
print("Lista:", brojevi)

# n-torka
n_torka = (10, 20, 30)
print("n-torka:", n_torka)

# Skup
skup = {1, 2, 2, 3}
print("Skup (bez duplikata):", skup)

# Rječnik
student = {"ime": "Ana", "godine": 22, "smjer": "Računarstvo"}
print("Rječnik:", student)
print("Pristup po ključu 'ime':", student["ime"])

# Iteracija
zbroj = 0
for x in brojevi:
    zbroj += x
print("Zbroj elemenata liste:", zbroj)

# List comprehension: kvadrati brojeva
kvadrati = [x**2 for x in brojevi]
print("Kvadrati:", kvadrati)
````


## 3. Kontrola toka i funkcije

`if/elif/else` i definiranje funkcija s `def`.


````python

def klasificiraj_broj(x):
    """Vrati tekstualnu oznaku za broj x (negativan, nula, pozitivan)."""
    if x < 0:
        return "negativan"
    elif x == 0:
        return "nula"
    else:
        return "pozitivan"

for val in [-2, 0, 7]:
    print(val, "->", klasificiraj_broj(val))
````


## 4. Rad s bibliotekama: `numpy` i `pandas`

- `numpy` za numeriku (vektori, statistika)  
- `pandas` za tablične podatke


````python

import numpy as np
import pandas as pd

# numpy: osnovne statistike
arr = np.array([1, 2, 3, 4, 5, 6])
print("Sredina:", np.mean(arr))
print("Medijan:", np.median(arr))
print("Std dev:", np.std(arr))
````


## 5. Kreiranje i učitavanje CSV datoteke

U praksi ćete češće čitati postojeće CSV datoteke. Ovdje za demonstraciju **kreiramo** mali skup podataka i spremamo ga u `podaci.csv`, zatim ga učitavamo s `pd.read_csv()`.


````python

# Kreiraj mali DataFrame i spremi kao CSV
df_demo = pd.DataFrame({
    "id": [1, 2, 3, 4, 5, 6],
    "spol": ["M", "Ž", "M", "Ž", "M", "Ž"],
    "dob": [23, 21, 25, 24, 29, 27],
    "visina_cm": [180, 165, 175, 170, 190, 160],
    "masa_kg": [75, 55, 68, 60, 95, 50]
})
csv_putanja = "podaci.csv"
df_demo.to_csv(csv_putanja, index=False)
csv_putanja
````

````python

# Učitavanje CSV-a
df = pd.read_csv(csv_putanja)

# Kratki pregled
display(df.head())
print("\nINFO:")
df.info()
print("\nDESCRIBE:")
display(df.describe())
````


## 6. Odabir, filtriranje, grupiranje

- Odabir stupaca i redaka  
- Filtriranje redaka uvjetima  
- Grupiranje i agregacije


````python

# Odabir stupaca
visine = df["visina_cm"]
print("Prosječna visina:", visine.mean())

# Filtriranje: osobe starije od 24
stariji_od_24 = df[df["dob"] > 24]
print("\nStariji od 24:")
display(stariji_od_24)

# Grupiranje: prosječna masa po spolu
prosjek_mase_po_spolu = df.groupby("spol")["masa_kg"].mean()
print("\nProsječna masa po spolu:")
display(prosjek_mase_po_spolu)
````


## 7. Vizualizacija (matplotlib)

````python

import matplotlib.pyplot as plt

# Histogram dobi
df["dob"].hist()
plt.xlabel("Dob")
plt.ylabel("Frekvencija")
plt.title("Histogram dobi")
plt.show()
````

````python

import matplotlib.pyplot as plt

# Scatter visina vs masa
plt.scatter(df["visina_cm"], df["masa_kg"])
plt.xlabel("Visina (cm)")
plt.ylabel("Masa (kg)")
plt.title("Visina vs. Masa — dijagram raspršenja")
plt.show()
````

# Funkcija `pd.read_csv()`

Funkcija `pd.read_csv()` iz biblioteke **pandas** služi za učitavanje CSV datoteka u **DataFrame**, što omogućava jednostavnu analizu i obradu tabličnih podataka.

---

## Osnovna sintaksa
```python
pd.read_csv("naziv_datoteke.csv")
```

Ova naredba učitava datoteku `naziv_datoteke.csv` u objekt tipa **DataFrame**.

---

## Najvažniji parametri

| Parametar | Značenje | Primjer |
|------------|-----------|----------|
| `filepath_or_buffer` | Putanja do datoteke | `"podaci.csv"` |
| `sep` | Separator stupaca (npr. zarez ili točka-zarez) | `sep=";"` |
| `decimal` | Znak koji se koristi za decimalne brojeve | `decimal=","` |
| `header` | Red koji sadrži nazive stupaca (0 = prvi red) | `header=0` |
| `names` | Ručno zadavanje naziva stupaca | `names=["Ime","Dob","Visina"]` |
| `index_col` | Određuje koji stupac koristiti kao indeks | `index_col="ID"` |
| `usecols` | Učitava samo odabrane stupce | `usecols=["Ime","Dob"]` |
| `dtype` | Definira tip podatka po stupcu | `dtype={"Dob": int}` |
| `na_values` | Vrijednosti koje se tretiraju kao prazne | `na_values=["-","n/a"]` |
| `encoding` | Kodiranje znakova (npr. UTF-8) | `encoding="utf-8"` |

---
